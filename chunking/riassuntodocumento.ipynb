```python
conn = sqlite3.connect(f"../RIASSUNTODOCUMENTI.db", check_same_thread=False)
cursor = conn.cursor()

cursor.execute("""
    CREATE TABLE IF NOT EXISTS document_chunks (
        filename TEXT,
        text_summary TEXT,
        PRIMARY KEY (filename)
    )
""")
conn.commit()
```

```python
def preparariassunto(DOC_SOURCE, DOC_SOURCE_MD):
    sql_data = []
    doc_summary = riassuntodocumento(DOC_SOURCE_MD)

    DOC_SOURCE = DOC_SOURCE.split("/")[-1].replace(".json", ".pdf")

    sql_data.append((DOC_SOURCE, doc_summary))

    cursor.executemany(
    "INSERT OR REPLACE INTO document_chunks (filename, text_summary) VALUES (?, ?)", 
    sql_data
    )
    conn.commit()

```

```python
base_folder = "../preprocessing/scratch"


json_file = ""
md_file = ""


for root, dirs, files in os.walk(base_folder):
    for file in files:
        if file.endswith(".json"):
            json_file = os.path.join(root, file)
        elif file.endswith(".md"):
            md_file = os.path.join(root, file)
    if json_file and md_file:
        #print(json_file)
        preparariassunto(DOC_SOURCE=json_file, DOC_SOURCE_MD=md_file)
```
